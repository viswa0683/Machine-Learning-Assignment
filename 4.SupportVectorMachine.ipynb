```python
# Pandas Library is imported into the file Multiple Regression Exercise 1
import pandas as pd
```

```python
# Variable l_dataset
# assign the read_csv function from pandas into the variable l_dataset
l_dataset = pd.read_csv("insurance_pre.csv")
```

```python
# Print the variable to display the data from 50_Startups.csv file
l_dataset
```

```python
# to convert text (categorical data) into numbers (0 and 1).
# converts categories into separate columns.
l_dataset2 =pd.get_dummies(l_dataset)
```

```python
l_dataset2
```

```python
# Here 0 and 1 is not converted . so we use int data type to convert into 0 and 1
l_dataset3= pd.get_dummies(l_dataset, dtype=int)
```

```python
# Now the data's from excel is aligned
l_dataset3
```

```python
# it print  all the column heading in array list
l_dataset3.columns
```

```python
# Now we split the input and output columns to the variable
independent=l_dataset3[['age', 'bmi', 'children', 'sex_female', 'sex_male','smoker_no', 'smoker_yes']]
```

```python
independent
```

```python
dependent = l_dataset3[["charges"]]
```

```python
dependent
```

```python
# Now we going to split , Training Set and Test Set
# Import the sklearn library, module of model_selection and train_test_split function into the Model creation
from sklearn.model_selection import train_test_split
```

```python
# it print the training input, testing input, training output and testing output
train_test_split (independent, dependent, test_size=0.30, random_state=0)
```

```python
X_train, X_test, y_train, y_test=train_test_split (independent, dependent, test_size=0.30,random_state=0)
```

```python
# # Standardisation is included after testing the Model
# from sklearn.preprocessing import StandardScaler
# l_sc= StandardScaler()
# X_train=l_sc.fit_transform(X_train)
# X_test=l_sc.transform(X_test)
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
y_test
```

```python
# Support Vector Machine
# Model creation
# l_reg = SVR(kernel="linear") 
# linear taking more time , so we use rbf

from sklearn.svm import SVR
l_reg = SVR(kernel="rbf" ,C=1.0) 

# Train the model 
l_reg.fit (X_train, y_train)
```

```python
l_reg.intercept_
```

```python
l_reg.n_support_
```

```python
l_reg.support_
```

```python
# Test the model	
y_pred = l_reg.predict (X_test)
```

```python
y_pred
```

```python
# Evaluation Method
from sklearn.metrics import r2_score
l_score=r2_score (y_test, y_pred)
l_score
```
